````python
import pandas as pd
import datetime as dt
import numpy as np
from pandas.tseries.offsets import MonthEnd

drivers_df = pd.DataFrame({
    "driver_id": [10, 8, 5, 7, 4, 1, 6],
    "join_date": [
        "2019-12-10",
        "2020-01-13",
        "2020-02-16",
        "2020-03-08",
        "2020-05-17",
        "2020-10-24",
        "2021-01-05"
    ]
})

rides_df = pd.DataFrame({
    "ride_id": [6, 1, 10, 19, 3, 13, 7, 17, 20, 5, 2, 11, 15, 12, 14],
    "user_id": [75, 54, 63, 39, 41, 52, 69, 70, 81, 57, 42, 68, 32, 11, 18],
    "requested_at": [
        "2019-12-09",
        "2020-02-09",
        "2020-03-04",
        "2020-04-06",
        "2020-06-03",
        "2020-06-22",
        "2020-07-16",
        "2020-08-25",
        "2020-11-02",
        "2020-11-09",
        "2020-12-09",
        "2021-01-11",
        "2021-01-17",
        "2021-01-19",
        "2021-01-27"
    ]
})

accepted_rides_df = pd.DataFrame({
    "ride_id": [10, 13, 7, 17, 20, 5, 2, 11, 15, 12, 14],
    "driver_id": [10, 10, 8, 7, 1, 7, 4, 8, 8, 8, 1],
    "ride_distance": [63, 73, 100, 119, 121, 42, 6, 37, 108, 38, 90],
    "ride_duration": [38, 96, 28, 68, 92, 101, 38, 43, 82, 34, 74]
})


drivers_df['join_date'] = pd.to_datetime(drivers_df['join_date'])
rides_df['requested_at'] = pd.to_datetime(rides_df['requested_at'])

month_df = pd.DataFrame({"year" : 2020, "month" : np.arange(1,13)})

month_df["last_date"] = (
    pd.to_datetime(
        month_df["year"].astype(str) + "-" + month_df["month"].astype(str) + "-01"
    ) + MonthEnd(1)
)

accepted_rides_driver_df = (
    rides_df.merge(accepted_rides_df, on ='ride_id')
    .loc[lambda d: (d['requested_at'] >= '2020-01-01') & (d['requested_at'] < '2021-01-01')]
    .assign(
        month = lambda d: d['requested_at'].dt.month
    )
    .groupby('month', as_index = False)['driver_id']
    .nunique()
    .rename(columns = {'driver_id' :'accepted_rides'})
)

active_driver_df =(
    month_df.merge(drivers_df, how = 'cross')
    .loc[lambda d: d['join_date'] <= d['last_date']]
    .groupby("month", as_index= False)['driver_id']
    .nunique()
    .rename(columns = {'driver_id' :'active_drivers'})
)
final_df = (
    active_driver_df.merge(accepted_rides_driver_df, how ='left', on ='month')
    .assign(
        working_percentage = lambda d: (d['accepted_rides'] *100.0/d['active_drivers']).round(2).fillna(0)
    )
    .sort_values('month')
    [['month','working_percentage']]
)

print(final_df)
````
